## Importacion de librerias

```python
from bs4 import BeautifulSoup   # Analiza y extrae información de HTML 
from datetime import datetime   # Manejo de fechas y horas
import requests                 # Permite hacer peticiones HTTP
import time                     # Utilizada para añadir pausas o medir tiempos
```

## Introduccion de scraping

```python
req = requests.get('http://books.toscrape.com/') # Hacer una petición GET al sitio web
src = req.text #Guardar el contenido HTML de la respuesta
print(src[:1000]) # Mostrar los primeros 1000 caracteres del contenido HTML
```

```python
soup = BeautifulSoup(src, 'lxml') #Convierte el HTML en un árbol de objetos que podemos navegar fácilmente
print(soup.prettify()[:1000])     #Agrega indentación y saltos de línea para que el HTML sea más entendible
```

```python
a_tags = soup.find_all("a") # Buscar todas las etiquetas <a> en el documento HTML
print(a_tags[:10])          # Imprime los primeros 10 resultados
```

```python
a_tags = soup.find_all("a") #Usar find_all() para obtener todas las etiquetas <a>
a_tags_alt = soup("a")      #Usar la notación abreviada soup("a") que es equivalente a find_all("a")
print(a_tags[0])            # Primer <a> usando find_all
print(a_tags_alt[0])        # Primer <a> usando la forma abreviada
```

```python
print(len(a_tags)) #Contar el número total de etiquetas <a> (hipervínculos) encontradas en la página
```

```python
selected = soup.select("div.side_categories") #Usamos select() con sintaxis de selectores CSS
selected[:5] #Mostrar los primeros 5 enlaces encontrados
```

```python
menu_links = soup.select("div.side_categories")   # Busca todos los <a> con clase "gh-a11y-skip-button__link"
first_link = menu_links[0]                    # Toma el primer enlace encontrado
print(first_link)                             # Muestra el HTML completo del primer enlace
print('Class: ', type(first_link))            # Imprime el tipo de dato (bs4.element.Tag)
```

```python
print(first_link.text)   # Imprime solo el texto visible dentro del enlace <a>
```

## Scraping para obtencion de datos de la pagina

```python
#Extraer categorías disponibles (están en un menú lateral)
categorias = soup.select("div.side_categories ul li ul li a")

print("Categorías encontradas:", len(categorias))

# Mostrar algunas
for i, cat in enumerate(categorias[:10], start=1):
    nombre = cat.text.strip()
    enlace = "http://books.toscrape.com/" + cat["href"]
    print(f"{i}. {nombre} -> {enlace}")
```

```python
#Extraer libros de la página principal
# Cada libro está en un artículo con clase 'product_pod'
libros = soup.select("article.product_pod")

print("Libros encontrados en la página principal:", len(libros))

for i, libro in enumerate(libros[:5], start=1):  # Mostramos solo 5 para ejemplo
    titulo = libro.h3.a["title"]
    precio = libro.select_one("p.price_color").text
    disponibilidad = libro.select_one("p.availability").text.strip()
    
    print(f"{i}. {titulo}")
    print(f"   Precio: {precio}")
    print(f"   Disponibilidad: {disponibilidad}\n")
```

```python
# Navegar a una categoría (ejemplo: 'Travel')
categoria_url = "http://books.toscrape.com/catalogue/category/books/travel_2/index.html"

response_cat = requests.get(categoria_url)
soup_cat = BeautifulSoup(response_cat.text, "html.parser")

libros_cat = soup_cat.select("article.product_pod")

print("Libros encontrados en la categoría 'Travel':", len(libros_cat))

for i, libro in enumerate(libros_cat, start=1):
    titulo = libro.h3.a["title"]
    precio = libro.select_one("p.price_color").text
    print(f"{i}. {titulo} -> {precio}")
```

```python
#Extraer información detallada de un libro
# Entramos en el enlace del primer libro de la categoría Travel
primer_libro_url = "http://books.toscrape.com/catalogue/its-only-the-himalayas_981/index.html"

response_libro = requests.get(primer_libro_url)
soup_libro = BeautifulSoup(response_libro.text, "html.parser")

# Extraer título, precio, disponibilidad y descripción
titulo = soup_libro.h1.text
precio = soup_libro.select_one("p.price_color").text
disponibilidad = soup_libro.select_one("p.availability").text.strip()
descripcion = soup_libro.select_one("div#product_description + p").text

print("Título:", titulo)
print("Precio:", precio)
print("Disponibilidad:", disponibilidad)
print("Descripción:", descripcion[:200], "...")
```
